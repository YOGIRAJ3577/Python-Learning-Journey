```python
x= np.arange(0,4*np.pi,0.1)
y= np.sin(x)
plt.title("sine wave form")

plt.plot(x,y)
plt.show() # not nesasary coz of %Matplotlib inline but we have to write if we dont write this line in start 
```

```python
x= np.arange(0,5*np.pi,0.1)
y_sin= np.sin(x)
y_cos= np.cos(x)


plt.subplot(2,1,1)
plt.plot(x,y_sin,"g--")
plt.title("sin wave form")
plt.subplot(2,1,2)
plt.plot(x,y_cos,"r--")
plt.title("cos wave form")
plt.show() 
```

Box Plot 

```python
# data= [np.random.normal(0,std,100) for std in range (1,4)]
# plt.boxplot(data, 'vertical',patch_artist=True)
data = [np.random.normal(0, std, 100) for std in range(1, 4)]

# Create boxplot
plt.boxplot(data, orientation='vertical', patch_artist=True)

plt.title('Box Plot Example')
plt.xlabel('Dataset')
plt.ylabel('Values')
```
